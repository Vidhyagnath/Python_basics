```python
import os

def longitude(long):
    if long>= -180 and long<= 180:
        return True
    else: 
        return False
def latitude(lat):
    if lat>=-90 and lat<= 90:
        return True
    else:
        return False
folder_name="PYTHON_GIS"
if not os.path.exists(folder_name):
    os.mkdir(folder_name)



while True:
    longi_tude=float(input("Enter Longitude:"))
    lati_tude=float(input("Enter Latitude:"))
    Place_name=input("Enter the name of place").upper()

    if longitude(longi_tude) and latitude(lati_tude):
        print("Valid Coordinates")

        folder_path=folder_name +"/Valid_cordinates.txt"

        with open(folder_path,'a') as file:
            file.write(f"Latitude:{lati_tude},Longitude:{longi_tude}, Place Name:{Place_name}\n")

        print(" Coordinates saved successfully!")

    else:
        print(" Invalid Coordinates! Please enter correct latitude and longitude.")
 
    choice=input("Do you want to enter another coordinate? yes/no").lower()

    if choice=="no":
        print("Exiting the programme")
        break


```
